```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum())
```

```python
categorical_columns = ["survived", "pclass", "sex", "embarked", "class"]

for column in categorical_columns:
    print(f"\n{'=' * 55}")
    print(f"Frequency analysis: {column}")
    print("=" * 55)

    counts = df[column].value_counts(dropna=False)
    proportions = df[column].value_counts(normalize=True, dropna=False)
    percentages = proportions * 100

    result = pd.DataFrame({
        "Frequency": counts,
        "Proportion": proportions.round(4),
        "Percentage": percentages.round(2)
    })

    display(result)
```

```python
for column in categorical_columns:
    counts = df[column].value_counts(dropna=False)

    plt.figure(figsize=(7, 4))
    counts.plot(kind="bar", edgecolor="black")
    plt.title(f"Frequency Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

```python
print("INTERPRETATION OF RESULTS")
print("-" * 55)

for column in categorical_columns:
    percentages = df[column].value_counts(normalize=True, dropna=False) * 100
    most_common = percentages.idxmax()
    highest_percentage = percentages.max()

    print(
        f"In '{column}', the most frequent category is "
        f"'{most_common}', representing {highest_percentage:.2f}% "
        "of the records."
    )

print("\nAnalysis completed successfully!")
```
